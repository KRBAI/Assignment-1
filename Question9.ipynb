```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Load an image of your choice
# Replace 'my_image.jpg' with any image you have
img = cv2.imread('/content/drive/MyDrive/Assignment1/taylor.jpg')

# 2. Define a Sharpening Kernel
# This kernel highlights edges by subtracting neighbors from the center
# Sum of kernel = 1 (keeps overall brightness the same)
kernel = np.array([[0, -1,  0],
                   [-1, 5, -1],
                   [0, -1,  0]])

# 3. Apply the kernel
sharpened = cv2.filter2D(img, -1, kernel)

# 4. Display
plt.figure(figsize=(10, 5))
# Convert BGR to RGB for matplotlib
plt.subplot(121), plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)), plt.title('Original')
plt.subplot(122), plt.imshow(cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB)), plt.title('Sharpened')
plt.show()
```
