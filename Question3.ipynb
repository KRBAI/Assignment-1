```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def equalize_hist_manual(img):
    # 1. Calculate the histogram
    hist, _ = np.histogram(img.flatten(), 256, [0, 256])

    # 2. Calculate the Cumulative Distribution Function (CDF)
    cdf = hist.cumsum()

    # 3. Normalize CDF to be our transformation function (0-255)
    # Formula: (L-1) * CDF / Total_Pixels
    cdf_normalized = (cdf * 255 / cdf.max()).astype('uint8')

    # 4. Map the original pixels to new values using the normalized CDF
    # This effectively does g = t[f]
    return cdf_normalized[img]

# Load grayscale image
img = cv2.imread('/content/drive/MyDrive/Assignment1/runway.png', 0)

# Apply our function
new_img = equalize_hist_manual(img)

# Display results
plt.figure(figsize=(10, 5))
plt.subplot(121), plt.imshow(img, cmap='gray'), plt.title('Original')
plt.subplot(122), plt.imshow(new_img, cmap='gray'), plt.title('Equalized')
plt.show()
```
