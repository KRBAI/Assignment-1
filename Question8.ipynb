```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/drive/MyDrive/Assignment1/emma_salt_pepper.jpg', 0)

if img is None:
    print("Error: Image not found.")
    exit()

# 8(a) Apply Gaussian Smoothing
# Uses a 5x5 kernel.
# Result: The noise usually gets blurry but stays visible.
gaussian_blur = cv2.GaussianBlur(img, (5, 5), 0)

# 8(b) Apply Median Filtering
# Uses a 5x5 kernel (ksize must be an odd integer like 3, 5, 7).
# Result: The noise should be completely removed.
median_filter = cv2.medianBlur(img, 5)

# --- Visualization ---
plt.figure(figsize=(15, 5))

# Original Noisy Image
plt.subplot(131)
plt.imshow(img, cmap='gray')
plt.title('Original (Salt & Pepper Noise)')
plt.axis('off')

# Gaussian Result
plt.subplot(132)
plt.imshow(gaussian_blur, cmap='gray')
plt.title('Gaussian Smoothing\n(Noise is blurred, not removed)')
plt.axis('off')

# Median Result
plt.subplot(133)
plt.imshow(median_filter, cmap='gray')
plt.title('Median Filtering\n(Best for this noise)')
plt.axis('off')

plt.show()
```
